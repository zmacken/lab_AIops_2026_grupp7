### Markdown Data Exploration
##### Utforska och analysera Markdown-filerna som samlats in med Firecrawl.

```python
from pathlib import Path
import pandas as pd
```

```python
# hitta alla Markdown-filer

data_path = Path("src/data")

md_files = list(data_path.glob("*.md"))

print(f"Antal Markdown-filer: {len(md_files)}")
```

```python
# visa de första fem filnamnen

for file in md_files[:5]:
    print(file.name)
```

```python
# Läs innehållet i en Markdown-fil

# Välj den första Markdown-filen
file_path = md_files[0]

# Läs innehållet
content = file_path.read_text(encoding="utf-8")

# Visa innehållet
print(content[:3000])
```
